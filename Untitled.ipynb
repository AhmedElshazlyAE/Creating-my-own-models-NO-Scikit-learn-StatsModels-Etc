```python
#
import pandas as pd
import numpy as np
import random
import math
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
def Z_score_standardization(X):
    return (X - X.mean()) / X.std()
```

```python
def loss(y, y_pred):
    return (1/len(y))*((y-y_pred)**2).sum()
```

```python
def Simple_Linear_Regression(X, y):
    b0 = random.random()
    b1 = random.random()
    X_scaled = Z_score_standardization(X)
    y_pred = b0 + b1*X_scaled
    return X_scaled, y, b0, b1
```

```python
def fit(model, alpha=0.01, itterations=1000, show_loss=False, loss_iteration=100):
    X = model[0]
    y = model[1]
    b0 = model[2]
    b1 = model[3]
    n = len(X)
    old_loss = math.inf
    loss_count = 0
    for i in range(itterations):
        y_pred = b0 + b1*X
        dX0 = (2/n) * (y_pred - y).sum()
        dX1 = (2/n) * ((y_pred - y) * X).sum()

        #trial step
        b0_t = b0 - alpha*dX0
        b1_t = b1 - alpha*dX1

        y_pred_t = b0_t + b1_t*X
        new_loss = loss(y, y_pred_t)
        
        if old_loss < new_loss:
            print(f"Stopping early at iteration {i}, loss increased to {new_loss}")
            break
            
        b0 = b0_t
        b1 = b1_t
        old_loss = new_loss
        loss_count += 1
        if show_loss and loss_count % loss_iteration == 0:
            print(f"- epoch {loss_count}, loss = {new_loss}")
        
   
    return b0, b1
```

```python
def predict(fitted_model, X):
    predictions = []
    for i in X:
        predictions.append(fitted_model[0] + fitted_model[1] * (i - X.mean()) / X.std())
    return pd.Series(predictions)
    
```

```python
area_train = np.linspace(10, 100, 50)

area_test = np.linspace(10, 100, 10)

price_train = 10 * area_train + np.random.normal(0, 20, size=area_train.shape)

price_test = 10 * area_test + np.random.normal(0, 35, size=area_test.shape)

model = Simple_Linear_Regression(area_train, price_train)
#Unscaled data diverged 
fitted = fit(model, alpha=0.0004,itterations=40000, show_loss=True, loss_iteration = 1000)
y_pred = predict(fitted, area_test)
```

```python
scatter = sns.scatterplot(x="area", y="price", data = pd.DataFrame({"area": area_test, "price": price_test}), color="red", label = "Observed")
line = plt.plot(area_test, y_pred, label = "Predict")
plt.legend()
```
