```python
class Optimizer:
    def __init__(self, lr):
        self.lr = lr

    def gradient_descent(self,B, dB):
        return B - self.lr * dB
```

```python
# Multiple Linear Regression
# y = b0 + b1x1 + b2x2 + ... + bnxn 
class Linear_Regression:
    n = 0 # Data Row Size
    b_count = 0 # Number of params 

    def __init__(self, lr=0.001, itterations=10000, optimizer="gradient_descent", patience=5):
        self.lr = lr
        self.itterations = itterations
        self.optimizer = optimizer
        self.patience = patience
        self.fit = Linear_Regression.Fit(self)
        
        
    def beta_sop(self, B, X): # Sum Of Products Of betas (y predictions)
        Y = np.zeros(self.n)
        for i in range(len(B)):
            Y += B[i] * X[i]
        return Y

    
    def Z_score_standardization(self, X, Data):
        return (Data - X.mean()) / X.std()

    def mse(self, y, y_pred):
        return (1 / self.n) * ((y - y_pred) ** 2).sum()

    def beta_dv(self, X, y, y_pred):
        dB = np.zeros(self.b_count)
        for i in range(self.b_count):
            dB[i] = (2/self.n) * ((y_pred - y) * X[i]).sum()
        return dB
        

    class Fit:
        def __init__(self, model):
            self.model = model
            self.b_count = self.model.b_count
            self.lr = self.model.lr
            self.itterations = self.model.itterations
            self.optimizer = self.model.optimizer
            self.patience = self.model.patience
            self.beta_sop = self.model.beta_sop
            self.Z_score_standardization = self.model.Z_score_standardization
            self.mse = self.model.mse
            self.beta_dv = self.model.beta_dv
            
        def __call__(self, X, y):
            self.X = X
            self.y = y
            self.n = len(y)
            
            self.scaled_X = np.vstack([np.ones(self.n), self.Z_score_standardization(self.X,self.X)])
            self.b_count = self.scaled_X.shape[0]
            self.B = np.zeros(self.b_count)

            self.model.b_count = self.b_count
            self.model.n = self.n
            
            solver = Optimizer(self.lr)
            loss_count = 0
            patience_idx = 0
            prev_loss = np.zeros(self.patience)
            best_loss = math.inf # Highest Possible Loss for early stopping
            best_prams = self.B
            print(self.X.shape)
            y_pred = self.beta_sop(self.B, self.scaled_X)
            
            for i in range(self.itterations):
                dB = self.beta_dv(self.scaled_X, self.y, y_pred)
                if self.optimizer == "gradient_descent":
                    self.B = solver.gradient_descent(self.B, dB)
            
                y_pred = self.beta_sop(self.B, self.scaled_X)
                loss = self.mse(self.y, y_pred)
            
                if(round(loss, 3) < round(best_loss, 3)):
                    best_prams = self.B
                    best_loss = loss
                    patience_idx = 0
                else:
                    patience_idx += 1
            
                if(patience_idx + 1 >= self.patience):
                    print("Early Stopping")
                    break

            print(best_loss)
            self.B = best_prams
            return self
            
        def predict(self, X):
            Y = np.zeros(len(X))
            X = np.vstack([np.ones(len(X)), self.Z_score_standardization(self.X, X)])
            for i in range(len(self.B)):
                Y += self.B[i] * X[i]
            return Y

        def _coeff(self):
            return self.B
```

```python
def gradient_descent(b, db, alpha):
    return b - alpha * db
```

```python
def adaptive_gradient(b, db, G, alpha):
    epsilon = 1 * 10**(-8)
    step = alpha / (G ** .5 + epsilon)
    return b - step * db
```

```python
def fit(model, alpha=0.01, itterations=1000, show_loss=False, loss_iteration=100, optimizer="gradient_descent"):
    X = model[0]
    y = model[1]
    b0 = model[2]
    b1 = model[3]
    n = len(X)
    old_loss = math.inf
    loss_count = 0
    G0 = 0
    G1 = 0
    LOSS = []
    for i in range(itterations):
        y_pred = b0 + b1*X
        db0 = (2/n) * (y_pred - y).sum()
        db1 = (2/n) * ((y_pred - y) * X).sum()

        #trial step
        if optimizer == "gradient_descent":
            b0_t = gradient_descent(b0, db0, alpha)
            b1_t = gradient_descent(b1, db1, alpha)
        elif optimizer == "adaptive_gradient":
            G0 += db0 ** 2
            G1 += db1 ** 2
            b0_t = adaptive_gradient(b0, db0, G0, alpha)
            b1_t = adaptive_gradient(b1, db1, G1, alpha)
        else:
            raise ValueError(f"No optimizer with name {optimizer} found")
        

        y_pred_t = b0_t + b1_t*X
        new_loss = loss(y, y_pred_t)
        
        if old_loss < new_loss:
            print(f"Stopping early at iteration {i}, loss increased to {new_loss}")
            break
            
        b0 = b0_t
        b1 = b1_t
        old_loss = new_loss
        loss_count += 1
        LOSS.append(new_loss)
        if show_loss and loss_count % loss_iteration == 0:
            print(f"- epoch {loss_count}, loss = {new_loss}")
        
   
    return {"b0": b0, "b1": b1, "loss": np.array(LOSS)}
```
